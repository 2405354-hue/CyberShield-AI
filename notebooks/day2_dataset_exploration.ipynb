# CyberShield AI - Day 2 Dataset Exploration

## Objective

Explore the CIC-IDS2017 dataset and understand the network traffic features used for intrusion detection.

## Dataset

The DDoS dataset contains:

- Rows: 225,745
- Columns: 79
- Classes: BENIGN and DDoS

### Class Distribution

- BENIGN: 97,718 (43.29%)
- DDoS: 128,027 (56.71%)

## Data Cleaning

The dataset was checked for missing values and infinite values.

The `Flow Bytes/s` feature contained 4 missing values caused by zero-duration flows. These values were replaced with 0.

Infinite values in `Flow Bytes/s` were also handled and replaced with 0.

After cleaning:

- Missing values: 0
- Infinite values: 0

## Feature Exploration

Important network traffic features were analyzed:

- Flow Duration
- Total Fwd Packets
- Total Backward Packets
- Total Length of Fwd Packets
- Total Length of Bwd Packets
- Flow Bytes/s

## Key Findings

The median values showed noticeable differences between BENIGN and DDoS traffic.

DDoS traffic showed:

- Higher median flow duration
- More forward and backward packets
- Much higher backward packet length
- Lower median Flow Bytes/s

These differences indicate that network-flow features can be useful for distinguishing normal traffic from DDoS attacks.

## Visualization

A boxplot of `Flow Bytes/s` was created to compare BENIGN and DDoS traffic.

A logarithmic scale was used because network traffic values have a very large range and contain many extreme values.

## Conclusion

The CIC-IDS2017 DDoS dataset contains useful network-flow characteristics that can help a machine learning model distinguish between normal and malicious traffic.

The cleaned dataset is ready for further feature preparation and model development.
